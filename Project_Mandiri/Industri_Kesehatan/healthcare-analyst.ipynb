```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
```

Perintah diatas untuk **mengubah data type** tanggal yang awalnya **object** menjadi **datetime**

```python
df['Length of Stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days
```

Perintah diatas merupakan perintah membuat **kolom lama rawat inap pasien** di rumah sakit

```python
df['Month'] = df['Date of Admission'].dt.month_name()
```

```python
df['Year'] = df['Date of Admission'].dt.year
```

Perintah diatas merupakan perintah **membuat kolom Month dan Year** sebagai bahan untuk analisis
